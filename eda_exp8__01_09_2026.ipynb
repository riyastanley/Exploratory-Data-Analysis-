```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/MNIST.csv')
```

```python
X_raw = df.drop(columns=['target']).values
y = df['target'].values
print(X_raw)
print(y)
```

```python
num_samples = X_raw.shape[0]
num_features = X_raw.shape[1]
unique_classes = np.unique(y)
num_classes = len(unique_classes)
print(num_samples)
print(num_features)
print(unique_classes)
print(num_classes)
```

```python
image_side = int(np.sqrt(num_features))
image_dims = (image_side, image_side)

print("--- Part A: Data Understanding ---")
print(f"Number of samples: {num_samples}")
print(f"Number of classes: {num_classes} (Labels: {list(unique_classes)})")
print(f"Image dimensions: {image_dims[0]} x {image_dims[1]} pixels ({num_features} total features)")
```

```python
plt.figure(figsize=(12, 5))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_raw[i].reshape(image_dims), cmap='gray')
    plt.title(f"Label: {y[i]}")
    plt.axis('off')
plt.suptitle("First 10 Sample Handwritten Digit Images", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
max_pixel_val = X_raw.max()
X_normalized = X_raw / float(max_pixel_val)

print("\n--- Part A: Preprocessing ---")
print(f"Original feature range: [{X_raw.min()}, {X_raw.max()}]")
print(f"Normalized feature range: [{X_normalized.min()}, {X_normalized.max()}]")
```

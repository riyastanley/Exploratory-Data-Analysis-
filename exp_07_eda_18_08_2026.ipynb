```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
#Descriptive Statistics
```

```python
mean = df["alcohol"].mean()
print(mean)
```

```python
median =df["alcohol"].median()
print(median)
```

```python
mode = df["alcohol"].mode()[0]
print(mode)
```

```python
df.alcohol.value_counts().nlargest(30).plot(kind='bar', figsize=(10,6))
plt.title("Number of quality by alcohol content")
plt.ylabel('Quality of wine')
plt.xlabel('Alcohol content')
```

```python
sns.FacetGrid(df).map(sns.distplot,"sulphates").add_legend()
```

```python
sns.FacetGrid(df).map(sns.distplot,"quality").add_legend()
```

```python
sns.boxplot(x="alcohol",data=df)
plt.show()
```

```python
#Bivariate Analysis
```

```python
plt.scatter(df["quality"], df["alcohol"])
plt.title("Scatter Plot for alcohol vs quality")
plt.xlabel("quality")
plt.ylabel("alcohol")
```

```python
sns.boxplot(x="quality",y="density",data=df)
plt.show()
```

```python
sns.boxplot(x="fixed acidity", y="density",data=df)
```

```python
#Multivariate Analysis
```

```python
sns.pairplot(df,vars = ['density', 'alcohol','quality'], kind="reg")
plt.show()
```

```python
sns.set(style="ticks", color_codes=True)
sns.pairplot(df,height=2,vars = ['density', 'alcohol','quality'], hue="pH")
plt.show()
```

```python
from scipy import stats
corr = stats.pearsonr(df["quality"], df["alcohol"])
print("p-value:\t", corr[1])
print("cor:\t\t", corr[0])
```

```python
correlation = df.corr(method='pearson', numeric_only=True)
correlation
```

```python
sns.heatmap(correlation,xticklabels=correlation.columns,
yticklabels=correlation.columns)
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(data=df, x="petal.length", hue="variety", kde=True, multiple="stack")
plt.title("Distribution of Petal Length by Variety")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df, x="sepal.length", y="sepal.width", hue="variety")
plt.title("Sepal Length vs Sepal Width")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x="variety", y="petal.width")
plt.title("Petal Width Variations Across Varieties")
plt.show()
```

```python
sns.pairplot(df, hue="variety", height=2.0)
plt.show()
```

```python
numeric_df = df.drop(columns=['variety'])
correlation_data = numeric_df.corr()

plt.figure(figsize=(6, 4.5))
sns.heatmap(correlation_data, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)

plt.title("Feature Correlation Heatgraph")
plt.tight_layout()
plt.show()
```
